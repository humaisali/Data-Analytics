*While dealing with the categorical data, there can be two cases such as ordinal and nominal. In the
nominal categorical data, there isn’t any kind of relationship in between the categories such as
category can be male or female and there isn’t any relationship or it cannot express any kind of  comparison etc. while in the ordinal category data, there is a clear relationship among categories or a kind of comparison such as the category data can be excellent, good, worst, average etc. The technique that deals with the nominal data is called as One-hot encoding while the technique deals with the ordinal data is called as ordinal encoding. Your task is to implement the ordinal encoding, label encoding and one-hot encoding. The implementation can be analyzed through different methods such as pandas and scikit-learn if possible.*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
```

```python
df=pd.read_csv('Raw.csv')
df.sample(5)
```

```python
ordinal_columns = [
    "Education_Level",
    "Satisfaction",
    "Performance"
]

nominal_columns = [
    "Gender",
    "City",
    "Department"
]
```

```python
ordinal_encoder = OrdinalEncoder(
    categories=[
        ["Bachelor's", "Master's", "PhD"],
        ["Low", "Medium", "High"],
        ["Poor", "Average", "Good", "Excellent"]
    ]
)

df_ordinal = df.copy()

df_ordinal[ordinal_columns] = ordinal_encoder.fit_transform(df_ordinal[ordinal_columns])

df_ordinal.sample(5)
```

```python
# Now let use the One Hot Endoder to encode the nominal columns

one_hot_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

encoded_nominal = one_hot_encoder.fit_transform(
    df[nominal_columns]
)

encoded_nominal_df = pd.DataFrame(
    encoded_nominal,
    columns=one_hot_encoder.get_feature_names_out(nominal_columns)
)

print("\nOne-Hot Encoded Data:")
print(encoded_nominal_df)
```
