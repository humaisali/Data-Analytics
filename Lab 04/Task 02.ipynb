**Task 02: Pretend your dataset has a variable/column having mixed numerical and categorical data such as
the rows are either numerical or categorical and not mixed in a row. Your task is to create two
columns which will extract the numerical and categorical data respectively**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
```

```python
df=pd.read_csv('Raw.csv')

df.head(10)
```

```python
# Non-numeric values become NaN

df["case_two_numeric"] = pd.to_numeric(df["case_two"], errors="coerce")

# Display the result
print(df[["case_two","case_two_numeric"]].sample(20))
```

```python
# Numeric values become NaN

df["case_two_categorical"] = df["case_two"].where(pd.to_numeric(df["case_two"], errors="coerce").isna())

# Display the result
print(df[["case_two","case_two_numeric","case_two_categorical"]].sample(20))
```
