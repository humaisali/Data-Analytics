```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
```

```python
df=pd.read_csv('Raw.csv')

df.sample(5)
```

```python
# case_one where there should be separate columns for the numerical and categorical data
df['case_one_numerical'] = df['case_one' ].str.extract('(\d+)',expand=False).astype(float)

print("Extracted Numerical Data")
print(df[["case_one", 'case_one_numerical' ] ].head())
```

```python
# case_one categorical column extraction
df['case_one_categorical' ] = df['case_one' ].str.replace("[0-9\s-]",'',regex=True)

print("Extracted Categorical Data")
print(df[['case_one', 'case_one_categorical']].head())
```

```python
# Filled the Numerical NaN = 0 and categorical NaN = 'Missing'
df['case_one_numerical'] = df['case_one_numerical' ].fillna(0)
df['case_one_categorical'] = df['case_one_categorical' ].fillna( 'Missing' )

print(df[['case_one','case_one_numerical', 'case_one_categorical' ]] . sample(10))
```
