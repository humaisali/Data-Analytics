*In many real-world datasets, date and time information is stored in text format, often with
inconsistent formats (e.g., 9/28/2024 10:30, 20240520, 2022-Jan-01, or even invalid strings like
Yesterday). Such inconsistencies make it difficult to perform analysis, time-based filtering, or
build models that rely on temporal features. Create combination of different date format in your
dataset or can download as well. The goal is to clean raw date-time values, convert them into a
standardized format, and extract useful features such as year, month, day, weekday, and hour for
further analysis or modeling. One of the pandas function that can help you is the “pd.to_datetime()”.
At the end display all the raw and clean date.*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
```

```python
df=pd.read_csv('Raw.csv')
```

```python
# Convert mixed date formats into DateTime
df["clean_date_time"] = pd.to_datetime(df["date_time_column"], errors="coerce", format="mixed")
```

```python
# Extract useful date-time features
df["year"] = df["clean_date_time"].dt.year
df["month"] = df["clean_date_time"].dt.month
df["day"] = df["clean_date_time"].dt.day
df["weekday"] = df["clean_date_time"].dt.day_name()
df["hour"] = df["clean_date_time"].dt.hour

# Standardize the date format for display
df["clean_date_time"] = df["clean_date_time"].dt.strftime("%Y-%m-%d %H:%M:%S")
```

```python
# Display raw and clean dates
print(df[["id","date_time_column","clean_date_time"]].to_string(index=False))
```
