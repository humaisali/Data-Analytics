```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the Dataset
df = pd.read_csv("D:\\1. Academics\\7th Semester Stuff\\Data Analytics Lab\\Lab 02\\Titanic-Dataset.csv")

df.sample(5)
```

```python
# Separate the Numerical and Categorical Variables

numerical_columns = ['Age', 'Fare', 'SibSp', 'Parch']

categorical_columns = ['Survived', 'Pclass', 'Sex', 'Embarked']

print("Numerical Variables:")
print(numerical_columns)

print("\nCategorical Variables:")
print(categorical_columns)
```

```python
# Describe the Numerical Variables

df[numerical_columns].describe()
```

```python
# Describe the Categorical Variables

df["Embarked"].value_counts()
```

```python
# Frequency Distribution of Categorical Variables Through Pie Chart

df["Embarked"].value_counts().plot.pie(autopct='%1.1f%%')
plt.show()

```

```python
# Box Plot for Numerical Variables to indentify the Outliers

df["Age"].plot.box()
plt.title("Box Plot for Age")
plt.ylabel("Age")
plt.show()
```

```python
# Visualize the Skewness of the Numerical Variables with KDE Plot

df[numerical_columns].skew().plot.kde()
plt.show()
```
