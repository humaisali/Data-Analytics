*Implement basic concept of the positional arguments in function*

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Humais", 21)
```

*Implement basic concept of the keyword arguments in function*

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(name="Humais", age=21)
```

*Implement the lambda function. Also compare it with the normal function*

```python
add = lambda a, b: a + b

result = add(10, 5)

print(result)
```
