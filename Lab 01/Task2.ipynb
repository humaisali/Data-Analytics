*Create an integer array of 5 elements and find out the properties such as shape, size, dimension, data type of elements, total bytes consumed by array and size in bytes of one array*

```python
arr = np.array([10, 20, 30, 40, 50])

# Shape

print("Shape of Array: ",arr.shape)

# Size
print("Size of Array: ",arr.size)

# Dimnesion
print("Dimension of Array: ",arr.ndim)

# Data type
print("Data type of Array: ",arr.dtype)

# Total Bytes
print("Total Bytes of Array: ",arr.nbytes)

# Item Size
print("Item Size of Array: ",arr.itemsize)
```

*Create a two dimensional array of your own choice and find out the dimension property, shape and print the elements using NumPy built-in properties only.*

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

# Dimnesion
print("Dimension of Array: ",arr.ndim)

# Shape
print("Shape of Array: ",arr.shape)

```
