*While dealing with the categorical data for the removal technique of the missing rows, the ratio of before and after removing rows should be slightly changed. Apply the removal of the missing data on any categorical data of the downloaded dataset and present the result in terms of ratio.*

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Building_Permits.csv")
```

```python
selected_column = 'Existing Use'
```

```python
# Check Missing Values in a Selected Categorical Column

missing_values = df[selected_column].isnull().sum()

missing_percentage = (missing_values / len(df)) * 100

print("Selected Categorical Column:", selected_column)

print("\nTotal Rows:")
print(len(df))

print("\nMissing Values:")
print(missing_values)

print("\nMissing Percentage:")
print(round(missing_percentage, 2), "%")
```

```python
# Ratio before removing missing rows

ratio_before = (df[selected_column].value_counts(normalize=True, dropna=False) * 100)

print("\nCategory Ratio Before Removal:")
print(ratio_before)
```

```python
# Remove rows containing missing values

new_df = df.dropna(subset=[selected_column])
```
