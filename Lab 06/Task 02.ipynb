Task 02: Use the scikit-learn class for the min-max scaling technique and visualize the feature before and after.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# Load dataset
dataset_path = "wine-clustering.csv"
df = pd.read_csv(dataset_path)

# Select numerical features
Numerical_features = [
    'Alcohol', 'Malic_Acid', 'Ash', 'Ash_Alcanity',
    'Magnesium', 'Total_Phenols', 'Flavanoids',
    'Nonflavanoid_Phenols', 'Proanthocyanins',
    'Color_Intensity', 'Hue', 'OD280', 'Proline'
]

df.head()
```

```python
df_MinMax = df.copy()
```

```python
# Create the scaler
scaler = MinMaxScaler()

# Fit and transform the selected features
df_MinMax[Numerical_features] = scaler.fit_transform(
    df[Numerical_features]
)

df_MinMax.head()
```

```python
# Single plot for both original and scaled data
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
df['Alcohol'].hist(bins=15)
plt.title("Before Min-Max Scaling")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
df_MinMax['Alcohol'].hist(bins=15)
plt.title("After Min-Max Scaling")
plt.xlabel("Scaled Alcohol")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
