Task 04: Implement the max absolute normalization scaling technique

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
dataset_path = "wine-clustering.csv"
df = pd.read_csv(dataset_path)

# Select numerical features
Numerical_features = [
    'Alcohol', 'Malic_Acid', 'Ash', 'Ash_Alcanity',
    'Magnesium', 'Total_Phenols', 'Flavanoids',
    'Nonflavanoid_Phenols', 'Proanthocyanins',
    'Color_Intensity', 'Hue', 'OD280', 'Proline'
]

df.head()
```

```python
# Copy the original dataset
df_MaxAbs = df.copy()

# Apply Max Absolute Scaling
df_MaxAbs[Numerical_features] = (
    df[Numerical_features]
    / df[Numerical_features].abs().max()
)

df_MaxAbs.head()
```

```python
# Single plot for both original and scaled data
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
df['Alcohol'].hist(bins=15)
plt.title("Before Min-Max Scaling")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
df_MaxAbs['Alcohol'].hist(bins=15)
plt.title("After Min-Max Scaling")
plt.xlabel("Scaled Alcohol")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import MaxAbsScaler

# Create scaler
scaler = MaxAbsScaler()

# Copy the dataset
df_MaxAbs_SK = df.copy()

# Scale numerical features
df_MaxAbs_SK[Numerical_features] = scaler.fit_transform(
    df[Numerical_features]
)

df_MaxAbs_SK.head()
```
