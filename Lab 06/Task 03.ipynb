Task 03: Implement the Mean Normalization Scaling Technique

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
dataset_path = "wine-clustering.csv"
df = pd.read_csv(dataset_path)

# Select numerical features
Numerical_features = [
    'Alcohol', 'Malic_Acid', 'Ash', 'Ash_Alcanity',
    'Magnesium', 'Total_Phenols', 'Flavanoids',
    'Nonflavanoid_Phenols', 'Proanthocyanins',
    'Color_Intensity', 'Hue', 'OD280', 'Proline'
]

df.head()
```

```python
# Create a copy of the original dataset
df_MeanNorm = df.copy()

# Apply Mean Normalization to numerical features
df_MeanNorm[Numerical_features] = (
    df[Numerical_features] - df[Numerical_features].mean()
) / (
    df[Numerical_features].max() - df[Numerical_features].min()
)

df_MeanNorm.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Original data
axes[0].hist(df['Alcohol'], bins=15)
axes[0].set_title('Before Mean Normalization')
axes[0].set_xlabel('Alcohol Value')
axes[0].set_ylabel('Frequency')

# Normalized data
axes[1].hist(df_MeanNorm['Alcohol'], bins=15)
axes[1].set_title('After Mean Normalization')
axes[1].set_xlabel('Normalized Alcohol')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
